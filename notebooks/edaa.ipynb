```python
# === RESET AND RECREATE FINAL DATA ===

# 1. Get numerical columns
numerical_columns = [col for col in X_train.columns 
                     if col not in categorical_columns and col not in ordinal_columns]

# 2. Take the parts (make sure indices are clean)
X_train_num = X_train[numerical_columns].reset_index(drop=True)
X_test_num  = X_test[numerical_columns].reset_index(drop=True)

X_train_onehot_df = X_encoded_df.reset_index(drop=True)
X_test_onehot_df  = X_test_encoded_df.reset_index(drop=True)

X_train_ordinal_df = X_train_ordinal_df.reset_index(drop=True)
X_test_ordinal_df  = X_test_ordinal_df.reset_index(drop=True)

# 3. Combine cleanly
X_train_final = pd.concat([X_train_num, X_train_onehot_df, X_train_ordinal_df], axis=1)
X_test_final  = pd.concat([X_test_num,  X_test_onehot_df,  X_test_ordinal_df],  axis=1)

print("Fixed X_train_final shape:", X_train_final.shape)
print("Fixed X_test_final shape: ", X_test_final.shape)
```

```python
scaler= StandardScaler()
```

```python
X_train_scaler=scaler.fit_transform(X_train_final)
X_test_scaler=scaler.transform(X_test_final)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pandas as pd
```

```python
models = {
    "1. Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "2. KNN": KNeighborsClassifier(n_neighbors=5),
    "3. Decision Tree": DecisionTreeClassifier(max_depth=6, random_state=42),
    "4. Random Forest": RandomForestClassifier(n_estimators=100, max_depth=8, random_state=42)
}
```

```python
results=[]
```

```python
y_predict=model.predict(X_test_scaler_df)
```

```python
results = []
for name,model in models.items():
    
    model.fit(X_train_scaler_df,y_train)          # Train

    y_pred=model.predict(X_test_scaler)
          # Predict on test
    acc = accuracy_score(y_test,y_pred)
   # acc = accuracy_score(y_test, y_pred)
    print(f"{name} Accuracy: {acc:.4f}")
    
    results.append({"Model": name, "Accuracy": acc})
    
    if acc > 0.85:
        print(classification_report(y_test, y_pred))
        print("-" * 60)

# Final comparison
results_df = pd.DataFrame(results)
print("\n=== Model Comparison ===")
print(results_df.sort_values(by="Accuracy", ascending=False))
```
